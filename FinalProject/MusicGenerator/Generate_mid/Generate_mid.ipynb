```python
import requests
import json
import time
import datetime
import os
import sys
import pandas as pd
import numpy as np
```

```python
import mido
from mido import Message, MidiFile, MidiTrack
```

```python
notes = []
def notes_append(note):
    notes.append(note)
```

```python
np.random.seed(int(time.time()))
for i in range(1, 101):
    vector = np.random.permutation([60, 62, 64, 65, 67, 69, 71, 72])
    notes = [
    (60, 120, 240),
    (62, 240, 240),
    (64, 180, 240),
    (65, 240, 240),
    (67, 300, 240),
    (69, 180, 240),
    (71, 240, 240),
    (72, 240, 240)
    ]

    # 讀取不同的 CSV 文件
    for j in range(1, 9):
        df = pd.read_csv(f'C:\\DL\\FinalProject\\MusicGenerator\\Generate_mid\\stock\\{j}_weekly.csv')
        df['Date'] = pd.to_datetime(df['Date'])
        df = df.set_index('Date')
        df = df.sort_index(ascending=True)
        
        # 從第二筆開始比較，如果平均股價上漲超過2%則紀錄下來
        for k in range(1, len(df)):
            if (df['Close'][k] - df['Close'][k-1]) / df['Close'][k-1] > 0.02:
                #創建time vector
                existtime = [30,60,120,180,240]
                # 隨機選擇一個start_time
                if existtime:
                    existtime = np.random.choice(existtime)
                else:
                    existtime =300
                print(existtime)
                notes.append((vector[j-1], existtime , existtime))

    # 打亂notes的順序
    notes = np.random.permutation(notes)
    
    # 創建一個MIDI文件和一個軌道
    mid = MidiFile()
    track = MidiTrack()
    mid.tracks.append(track)

    # 添加MIDI頭信息
    track.append(Message('program_change', program=12, time=0))

    # 添加音符到MIDI軌道
    for note in notes:
        pitch, start_time, duration = note
        # 添加Note On消息
        track.append(Message('note_on', note=pitch, velocity=64, time=start_time))
        # 添加Note Off消息
        track.append(Message('note_off', note=pitch, velocity=64, time=start_time + duration))

    # 保存MIDI文件
    save_path = f'C:\\DL\\FinalProject\\MusicGenerator\\pool\\output{i}.mid'
    mid.save(save_path)
```
